Весь аналіз даних будується на тому, що ми перевіряємо певні гіпотези та робимо висновки і рекоемндації на основі своїх досліджень.

```python
print(df[(df.City == 'Yangon')]['gross income'].mean(), df[(df.City == 'Mandalay')]['gross income'].mean())
```

In Mandalay City, our mean gross income is higher.

```python
df.columns
```

Hypothesis 1: Higher average check
We assume that Mandalay generates higher gross income because customers
spend more money per single purchase compared to Yangon.

```python
print(df[df.City == 'Yangon'].Total.mean(), df[df.City == 'Mandalay'].Total.mean())
```

The hypothesis was confirmed - in the city with higher gross income (Mandalay),
customers also spend more per single purchase.
A higher average check directly contributes to higher gross income.
Business decision: To increase profit in Yangon, it is worth stimulating higher average spending - for example, through "spend X - get a discount" promotions
or cross-selling strategies (suggesting related products at checkout).

Hypothesis 2 : Higher total quantity of items sold

We assume that Mandalay generates higher gross income because customers
buy more items in total compared to Yangon.

```python
print(df[df.City == 'Yangon'].Quantity.sum(), df[df.City == 'Mandalay'].Quantity.sum())
```

The hypothesis was not confirmed - Yangon actually sold
more items in total (1859 vs 1820), yet Mandalay has higher gross income.
This means quantity of items sold is not the main driver of gross income.

Business decision: Since selling more items does not necessarily lead to
higher income, the focus should be on promoting higher-priced products
rather than simply increasing the number of items sold.

Hypothesis 3: Higher number of transactions

We assume that Mandalay generates higher gross income because it has more purchases
compared to Yangon.

```python
df.query('City == "Yangon" | City == "Mandalay"').City.value_counts()
```

The hypothesis was not confirmed - Yangon actually has more
transactions (340 vs 332), yet Mandalay has higher gross income.
This means the number of purchases is not the main driver of gross income.

Business decision: Since transaction count is not the key factor,
the focus should be on increasing the average check or promoting
higher-margin product categories rather than simply attracting more customers.
