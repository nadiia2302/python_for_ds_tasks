Гіпотеза 2. У яку годину дня спостерігається найбільша кількість покупок та найбільша сума продажів?

```python
df['Time'] = pd.to_datetime(df['Time'])
```

```python
df['hours_from_timestamps'] = df['Time'].dt.hour
```

```python
df.head()
```

```python
pivot_by_quantity_and_total = df.pivot_table(index = 'hours_from_timestamps', values = ['Total', 'Quantity'], aggfunc={'Total':'sum','Quantity':'size'})
```

```python
pivot_by_quantity_and_total.sort_values(['Total', 'Quantity'], ascending=False)
```

У 19, 13, 10 - х годинах клієнти витрачають найбільше грошей на покупки
