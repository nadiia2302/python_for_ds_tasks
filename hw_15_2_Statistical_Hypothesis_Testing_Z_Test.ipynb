```python
import numpy as np
```

```python
sample_mean = np.mean(data)
population_mean = 20
population_std = np.std(data, ddof = 1)
sample_size = len(data)
alpha = 0.05
```

2) Формування гіпотези

Гіпотези

H₀: μ = 20 (продажі не кращі за середню)
H₁: μ > 20 (продажі кращі за середню)
Тип тесту: правосторонній (one-tailed)

Чому саме правосторонній тест?

Треба переевірити чи вийшло у нас статистично значущо підвищити середній чек за липень 2025?

```python
SE = population_std/np.sqrt(sample_size)
```

```python
from scipy import stats
```

```python
z_critical = stats.norm.ppf(1-alpha)
```

```python
z_statistic = (sample_mean - population_mean)/SE
```

```python
p_value = 1 - stats.norm.cdf(z_statistic)
```

Метод 1: порівняння Z-статистики з критичним значенням

```python
print('\n' + "="*60)
print("Порівняння z-статистики з критичним значенням")
print("="*60 + '\n')
if (z_statistic > z_critical):
  print(f"Висновок: z-статистика ({z_statistic:.3f}) ПЕРЕВИЩУЄ критичне значення ({z_critical:.3f})")
  decision = 'Відхиляємо H0'
else:
  print(f"Висновок: z-статистика ({z_statistic:.3f}) НЕ ПЕРЕВИЩУЄ критичне значення ({z_critical:.3f})")
  decision = 'Не відхиляємо H0'
print(f'{decision}')
```

Метод 2: порівняння p-value з рівнем значущості (alpha):


```python
print('\n'+'='*60)
print("Порівняння p-value з рівнем значущості")
print('\n')
if p_value < alpha:
  print(f"Висновок: p-value ({p_value:.3f}) МЕНШЕ, ніж рівень значущості alpha ({alpha:.3f})")
  decision = 'Відхиляємо H0'
else:
  print(f"Висновок: p-value ({p_value:.3f}) НЕ МЕНШЕ, ніж рівень значущості alpha ({alpha:.3f})")
  decision = 'Не відхиляємо H0'
print(f'{decision}')
```

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats import weightstats as stests
```

```python
print("=" * 60)
print("ВИКОРИСТАННЯ STATSMODELS ДЛЯ Z-ТЕСТУ:")
print("=" * 60 + '\n')

np.random.seed(42)
sample_data = np.random.normal(sample_mean, population_std, sample_size)

# корекція середнього значення як початкового
sample_data = sample_data - np.mean(sample_data) + sample_mean
```

```python
sample_data
```

```python
np.mean(sample_data)
```

```python
print("="*60)
print("Виконання z-test:")
print("="*60)

# Виконуємо правосторонній z-test
z_stat, p_val = stests.ztest(sample_data, value = population_mean, alternative='larger')

print(f"Результати z-test:")
print(f"Z-статистика: {z_stat:.3f}")
print(f"P-value: {p_val:.3f}")

print("\n"+"="*60)
print("Інтерпретація результатів:")
print("="*60)

if p_val < alpha:
  print(f"Відхиляємо H0")
  print(f"p-value ({p_val:.3f}) < alpha ({alpha})")
  print("Вийшло статистично значущо підвищити середній чек за липень 2025")
else:
  print(f"Не відхиляємо H0")
  print(f"p-value ({p_val:.3f}) >= alpha ({alpha})")
  print("Не вийшло статистично значущо підвищити середній чек за липень 2025")

print("\n"+"="*60)
print("ПОРІВНЯННЯ З РУЧНИМИ РОЗРАХУНКАМИ:")
print("=" * 60)
print(f"Z-статистика:")
print(f"Ручний розрахунок z-statistic: {z_statistic:.3f}")
print(f"Statsmodels z-statistic:       {z_stat:.3f}")
print(f"Різниця z-statistic:           {abs(z_statistic - z_stat):.3f}")
print()
print(f"Ручний розрахунок p-value: {p_value:.3f}")
print(f"Statsmodels p-value:       {p_val:.3f}")
print(f"Різниця p-value:           {abs(p_value - p_val):.3f}")
print()
print("Примітка: Невелика різниця пояснюється тим, що ми")
print("використовуємо згенеровані дані, а не точне середнє 20")
```
