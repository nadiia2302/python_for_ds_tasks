**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

1.Формування гіпотези:

H0: нью0 = 123.7 (середній час до покупки користувачів не збільшився за середню)

H1: нью > 123.7 (середній час до покупки користувачів збільшився за середню)

Тип тесту правосторонній (one-tailed)

Чому саме правосторонній тест?

E-commerce компанія підозрює, що середній час до покупки користувачів збільшився. Тому перевіряєму чи середній час збільшився, а не просто відрізняється



```python
import numpy as np
from scipy import stats
times_sample = np.array([128, 135, 121, 142, 126, 151, 123])
n = len(times_sample)
mean_sample = np.mean(times_sample)
s = np.std(times_sample, ddof = 1)
```

3.Тип тесту t-test (для малих вибірок)

```python
mu0 = 123.7
alpha = 0.1

# Обчислюємо t-статистику вручну
t_stat = (mean_sample - mu0)/(s/np.sqrt(n))

# Критичне значення для лівостороннього тесту
df = n-1
t_crit = stats.t.ppf(1-alpha, df)

p_value = 1 - stats.t.cdf(t_stat, df)
```

```python
print(f"При alpha = {alpha}")
print(f"t-статистика: {t_stat:.3f}")
print(f"t-критичне: {t_crit:.3f}")
print(f"p-value: {p_value:.3f}")
print(f"alpha: {alpha}")

print("-"*60)
print(f"Перевірка гіпотези за допомогою p-value і при alpha = {alpha}:")
print("-"*60)
if (p_value < alpha):
  print("Відхилити H0 гіпотезу: середній час до покупки користувачів збільшився")
else:
  print("Не відхилити H0 гіпотезу: доказів недостатньо")
```

Після перевірки гіпотези, відхиляємо H0 гіпотезу: тобто середній час до покупки у користувачів збільшився при alpha = 0.1

```python

alpha2 = 0.05
print(f"При alpha = {alpha2}")
t_crit2 = stats.t.ppf(1-alpha2,df)
print(f"t-статистика: {t_stat:.3f}")
print(f"t-критичне: {t_crit2:.3f}")
print(f"p-value: {p_value:.3f}")
print("-"*60)
print(f"Перевірка гіпотези за допомогою p-value і при alpha = {alpha2}:")
print("-"*60)
if (p_value < alpha2):
  print("Відхилити H0 гіпотезу: середній час до покупки користувачів збільшився")
else:
  print("Не відхилити H0 гіпотезу: доказів недостатньо")

```

Після перевірки гіпотези, ми відхиляємо H0 гіпотезу: тобто середній час до покупки у користувачів збільшився при alpha = 0.05

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

H0: p = 0.23 (Не більше ніж 23% доросли упізнавали логотип компанії)

Ha: p > 0.23 (Більше ніж 23% дорослих упізнавали логотип компанії)

alpha = 0.03

x = 311

n = 1200

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
```

```python
p = 0.23
x = 311
n  = 1200
alpha = 0.03
stat, pval = proportions_ztest(x, n , p, alternative = "larger")
confint = proportion_confint(x, n, alpha, method = 'normal')
print("Тест для однієї пропорції")
print(f"p-value: {pval:.2f}")
print(f"Z-статистика: {stat:.2f}")
if (pval < alpha):
  print(f"Відхилити гіпотезу H0. На рівні значущості {alpha} можно стверджувати, що більше ніж {p*100:.0f}% дорослих упізнають логотип компанії")
else:
  print(f"Не відхилити гіпотезу H0. На рівні значущості {alpha} не можно стверджувати, що більше ніж {p*100:.0f}% дорослих упізнають логотип компанії")
print(f"97% довірчий інтервал:: {confint}")

```

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

97% довірчий інтервал (0.231, 0.286)

З ймовірністю 97% справжня частка лежить в інтервалі від 23.1% до 28.6%

Гіпотетичне значення лежить перед довірчим інтервалом -> відхилити H0
